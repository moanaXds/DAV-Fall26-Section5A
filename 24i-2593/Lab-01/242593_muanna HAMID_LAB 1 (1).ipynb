```python
import pandas as pd
import numpy as np

df = pd.read_csv('adult.csv')

df.info()
```

```python
df.head()
```

```python
df.isnull().sum()
```

```python
df.replace('?', np.nan, inplace=True)
df.isnull().sum()
```

## Task 2 

```python
missing_count = df.isnull().sum().sort_values(ascending=False)
missing_pct = (df.isnull().sum() / len(df) * 100).round(2).sort_values(ascending=False)

missing_report = pd.DataFrame({'missing_count': missing_count, 'missing_pct': missing_pct})
missing_report = missing_report[missing_report['missing_count'] > 0]
missing_report
```

## Task 3 

```python
df['workclass'] = df['workclass'].fillna('Unknown')
df['occupation'] = df['occupation'].fillna('Unknown')
df['native.country'] = df['native.country'].fillna(df['native.country'].mode()[0])

df[['workclass', 'occupation', 'native.country']].isnull().sum()
```

## Task 4 

```python
exact_dupes = df.duplicated().sum()
exact_dupes
```

```python
dupes_ignoring_label = df.drop(columns=['income']).duplicated().sum()
dupes_ignoring_label
```

```python
df = df.drop_duplicates()
df.shape
```

## Task 5 

```python
df['education'].unique()
```

```python
df['marital.status'].unique()
```

```python
df['native.country'].unique()
```

```python
for col in ['education', 'marital.status', 'native.country']:
    before = df[col].unique()
    
    df[col] = df[col].str.strip()
    after = df[col].unique()
    
    changed = set(before) - set(after)
    
    print(col, '-> stripped entries changed:', changed)
```

## Task 9

```python
import sqlite3

raw_df = pd.read_csv('adult.csv')

conn = sqlite3.connect('adult_income.db')
raw_df.to_sql('adult_income', conn, if_exists='replace', index=False)

query = "SELECT * FROM adult_income WHERE age > 30;"
sql_df = pd.read_sql_query(query, conn)
sql_df.head()
```

```python
print('CSV-loaded shape:', raw_df[raw_df['age'] > 30].shape)
print('SQL-extracted shape:', sql_df.shape)

print('Match:', raw_df[raw_df['age'] > 30].shape == sql_df.shape)

conn.close()
```

# Task 10  Data Profiling Summary Report

## Dataset Overview

The UCI Adult Income dataset has 32561 rows and 15 columns

It contains information about people such as age education job marital status and working hours

The goal is to predict whether a person earns more than 50K per year

## Quality Issues

Some missing values were hidden as question marks

Occupation had 1843 missing values

Workclass had 1836 missing values

Native country had 583 missing values

Some exact duplicate rows were also found

No extra spaces were found but whitespace was still removed as a safety step

## Key Findings

About 76 percent of people earn 50K or less while 24 percent earn more than 50K

Prof specialty is the most common occupation

The dataset has more males than females

The average age is about 39 years

Most people work around 40 hours per week

Education and education num match perfectly

## Cleaning Done

Question marks were changed to missing values

Missing workclass and occupation values were changed to Unknown

Missing native country values were filled with United States

Exact duplicate rows were removed

Rows with the same details but different income were kept

Extra spaces were removed from important categorical columns
