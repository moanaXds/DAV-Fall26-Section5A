```python
def contrast_ratio(c1, c2):

    def luminance(rgb):
        rgb = [x / 255 for x in rgb]

        rgb = [
            ((x + 0.055) / 1.055) ** 2.4 if x > 0.04045 else x / 12.92
            for x in rgb
        ]

        return (
            0.2126 * rgb[0] +
            0.7152 * rgb[1] +
            0.0722 * rgb[2]
        )

    l1 = luminance(c1)
    l2 = luminance(c2)

    lighter = max(l1, l2)
    darker = min(l1, l2)

    return (lighter + 0.05) / (darker + 0.05)
```

```python
c1 = (102, 194, 165)
c2 = (252, 141, 98)

ratio = contrast_ratio(c1, c2)

print("Contrast ratio:", round(ratio, 2))

if ratio >= 4.5:
    print("PASS: Meets WCAG AA")
else:
    print("FAIL: Below WCAG AA")
```

```python
profit_data = dataset.groupby('Sub-Category')['Profit'].sum().reset_index()
profit_data['Color'] = profit_data['Profit'].apply(
    lambda x: 'red' if x < 0 else 'green'
)

plt.figure(figsize=(12, 6))

plt.bar(
    profit_data['Sub-Category'],
    profit_data['Profit'],
    color=profit_data['Color']
)

plt.axhline(0, color='black', linewidth=1)

plt.title('Profit by Sub-Category')
plt.xlabel('Sub-Category')
plt.ylabel('Profit')

plt.xticks(rotation=45)
plt.show()
```
